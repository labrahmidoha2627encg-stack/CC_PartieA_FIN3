## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Régression linéaire (Linear Regression). **Famille :** Régression supervisée. **Objectif :** Prédire une valeur quantitative par une combinaison linéaire des variables.

**Intuition et fonctionnement.** Le modèle reçoit X et y, estime les coefficients par moindres carrés, puis calcule une prédiction pour chaque nouvelle observation.

**Formule ou critère essentiel :**  
`ŷᵢ = β₀ + Σⱼ βⱼxᵢⱼ ; J(β)=1/(2n)Σᵢ(yᵢ−ŷᵢ)²`

**Définition des symboles :** ŷᵢ : prédiction ; β₀ : constante ; βⱼ : coefficients appris ; xᵢⱼ : valeur de la variable j ; yᵢ : valeur réelle ; n : nombre d'observations.  
**Hyperparamètres et paramètres appris :** fit_intercept ; paramètres appris : β₀ et βⱼ.

**Deux avantages :**
- Simple et rapide.
- Coefficients relativement interprétables.

**Deux limites :**
- Ne capture pas spontanément les relations non linéaires.
- Sensible aux valeurs extrêmes et à la colinéarité.

### Schéma explicatif

Le schéma original ci-dessous représente le cheminement conceptuel de l'algorithme : données/état en entrée, opération d'apprentissage, puis sortie. Les étiquettes décrivent le mécanisme spécifique de cette méthode.

### Contextes d'utilisation

**Cas d'usage 1 :** Prévoir le chiffre d'affaires selon les dépenses commerciales (gestion).  
**Cas d'usage 2 :** Estimer les coûts logistiques selon distance, volume et carburant.

**Données et prérequis :** Régression supervisée. Il faut disposer de variables cohérentes avec la tâche, traiter les valeurs manquantes, choisir un encodage adapté et, lorsque la méthode dépend des distances ou des marges, mettre les variables à l'échelle. Les hyperparamètres doivent être réglés sur validation, jamais sur le test final.

**Cas peu approprié :** Classer un client en défaut/non-défaut : la cible est une catégorie, pas une quantité continue.  
**Métriques adaptées :** MAE, RMSE et R².

### Analyse de l'expérience Python

Le code utilise Xr_ent/Xr_test et une cible numérique synthétique ; LinearRegression() apprend une droite/hyperplan et affiche la MAE. Cela illustre l'ajustement et l'erreur absolue, mais ne fournit aucun résultat valable sur le dataset de crédit et ne permet pas d'évaluer une classification.



```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(10, 2.4))
ax.set_xlim(0, 3); ax.set_ylim(0, 1); ax.axis("off")
etapes = ['Variables X', 'Combinaison linéaire', 'Valeur prédite ŷ']
for j, texte in enumerate(etapes):
    ax.text(j+0.5, 0.52, texte, ha="center", va="center", wrap=True,
            bbox=dict(boxstyle="round,pad=0.6", facecolor="#e8eef7", edgecolor="#24476b"))
    if j < 2:
        ax.annotate("", xy=(j+0.93,0.52), xytext=(j+0.67,0.52),
                    arrowprops=dict(arrowstyle="->", lw=1.8))
ax.set_title("1. Régression linéaire (Linear Regression) — schéma de principe", fontsize=12)
plt.show()
```

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Ridge (Ridge Regression). **Famille :** Régression supervisée pénalisée L2. **Objectif :** Réduire l'amplitude des coefficients pour stabiliser une régression.

**Intuition et fonctionnement.** Le modèle calcule une prédiction linéaire et minimise l'erreur quadratique augmentée d'une pénalité sur les carrés des coefficients.

**Formule ou critère essentiel :**  
`J(β)=||y−Xβ||²₂/n + αΣⱼβⱼ²`

**Définition des symboles :** X : matrice des variables ; y : cible quantitative ; βⱼ : coefficients appris ; α : intensité de régularisation.  
**Hyperparamètres et paramètres appris :** alpha ; fit_intercept ; paramètres appris : coefficients β.

**Deux avantages :**
- Stabilise les coefficients en présence de variables corrélées.
- Réduit le surapprentissage par régularisation.

**Deux limites :**
- Ne supprime généralement pas les variables.
- Le choix d'alpha influence fortement le compromis biais-variance.

### Schéma explicatif

Le schéma original ci-dessous représente le cheminement conceptuel de l'algorithme : données/état en entrée, opération d'apprentissage, puis sortie. Les étiquettes décrivent le mécanisme spécifique de cette méthode.

### Contextes d'utilisation

**Cas d'usage 1 :** Prévoir des revenus avec indicateurs macroéconomiques corrélés (finance).  
**Cas d'usage 2 :** Prévoir les dépenses d'approvisionnement avec plusieurs facteurs liés.

**Données et prérequis :** Régression supervisée pénalisée L2. Il faut disposer de variables cohérentes avec la tâche, traiter les valeurs manquantes, choisir un encodage adapté et, lorsque la méthode dépend des distances ou des marges, mettre les variables à l'échelle. Les hyperparamètres doivent être réglés sur validation, jamais sur le test final.

**Cas peu approprié :** Une cible catégorielle de défaut, car Ridge est une régression et non un classifieur.  
**Métriques adaptées :** MAE, RMSE, R².

### Analyse de l'expérience Python

Le code standardise les entrées, ajuste Ridge(alpha=1.0) et affiche MAE sur un exemple de régression synthétique. Il montre l'effet de la pénalisation L2, mais aucune valeur imprimée ne doit être attribuée à la base de crédit.



```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(10, 2.4))
ax.set_xlim(0, 3); ax.set_ylim(0, 1); ax.axis("off")
etapes = ['X standardisé', 'Erreur + pénalité L2', 'Coefficients réduits']
for j, texte in enumerate(etapes):
    ax.text(j+0.5, 0.52, texte, ha="center", va="center", wrap=True,
            bbox=dict(boxstyle="round,pad=0.6", facecolor="#e8eef7", edgecolor="#24476b"))
    if j < 2:
        ax.annotate("", xy=(j+0.93,0.52), xytext=(j+0.67,0.52),
                    arrowprops=dict(arrowstyle="->", lw=1.8))
ax.set_title("2. Ridge (Ridge Regression) — schéma de principe", fontsize=12)
plt.show()
```

## 3. Lasso

### Présentation

**Nom :** Lasso (Lasso Regression). **Famille :** Régression supervisée pénalisée L1. **Objectif :** Effectuer une régression tout en pouvant annuler certains coefficients.

**Intuition et fonctionnement.** Après mise à l'échelle, le modèle minimise l'erreur quadratique et la somme des valeurs absolues des coefficients ; certains deviennent exactement nuls.

**Formule ou critère essentiel :**  
`J(β)=||y−Xβ||²₂/(2n)+αΣⱼ|βⱼ|`

**Définition des symboles :** βⱼ : coefficient appris ; α : intensité de la pénalité L1 ; n : nombre d'observations.  
**Hyperparamètres et paramètres appris :** alpha ; max_iter ; paramètres appris : βⱼ.

**Deux avantages :**
- Effectue une sélection intégrée de variables.
- Produit un modèle parfois plus parcimonieux.

**Deux limites :**
- Peut sélectionner arbitrairement une variable parmi des variables corrélées.
- Résultats sensibles à alpha et à l'échelle.

Le schéma original ci-dessous représente le cheminement conceptuel de l'algorithme : données/état en entrée, opération d'apprentissage, puis sortie. Les étiquettes décrivent le mécanisme spécifique de cette méthode.

### Contextes d'utilisation

**Cas d'usage 1 :** Sélectionner des indicateurs financiers pour prévoir une marge.  
**Cas d'usage 2 :** Identifier les facteurs associés aux coûts de transport.

**Données et prérequis :** Régression supervisée pénalisée L1. Il faut disposer de variables cohérentes avec la tâche, traiter les valeurs manquantes, choisir un encodage adapté et, lorsque la méthode dépend des distances ou des marges, mettre les variables à l'échelle. Les hyperparamètres doivent être réglés sur validation, jamais sur le test final.

**Cas peu approprié :** Prédire directement une classe binaire avec LassoRegressor plutôt qu'un classifieur adapté.

**Nom :** Régression polynomiale (Polynomial Regression). **Famille :** Régression supervisée. **Objectif :** Modéliser une relation courbe en ajoutant des puissances et interactions des variables.

**Définition des symboles :** x : variable d'entrée ; d : degré choisi ; β : coefficients appris ; ŷ : prédiction.  
**Hyperparamètres et paramètres appris :** degree ; include_bias ; alpha du Ridge éventuellement utilisé.

**Deux avantages :**
- Représente des courbes simples.
- S'appuie sur des modèles linéaires connus.

**Deux limites :**
- Le nombre de termes augmente vite avec les variables et le degré.
- Un degré élevé peut surapprendre et extrapoler mal.

**Cas peu approprié :** Un problème de classification à nombreuses variables et interactions complexes sans contrôle de complexité.

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Arbre de régression (Decision Tree Regressor). **Famille :** Régression supervisée. **Objectif :** Prédire une valeur numérique en partitionnant l'espace des variables en régions.

**Intuition et fonctionnement.** À chaque nœud, l'arbre choisit une variable et un seuil qui réduisent l'erreur quadratique ; les feuilles prédisent une moyenne locale.

**Formule ou critère essentiel :**  
`J_m=Σᵢ∈m(yᵢ−ȳ_m)²`

**Définition des symboles :** m : feuille ; yᵢ : valeur réelle ; ȳ_m : moyenne de la cible dans la feuille ; seuils et variables de coupure appris.  
**Hyperparamètres et paramètres appris :** max_depth ; min_samples_leaf ; random_state.

**Deux avantages :**
- Capture les relations non linéaires et interactions.
- Règles de décision visualisables.

**Deux limites :**
- Instable aux variations de données.
- Profondeur excessive : surapprentissage.

### Schéma explicatif

Le schéma original ci-dessous représente le cheminement conceptuel de l'algorithme : données/état en entrée, opération d'apprentissage, puis sortie. Les étiquettes décrivent le mécanisme spécifique de cette méthode.

### Contextes d'utilisation

**Cas d'usage 1 :** Estimer les pertes financières attendues (finance).  
**Cas d'usage 2 :** Prévoir le délai ou coût de livraison.

**Données et prérequis :** Régression supervisée. ## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Q-learning. **Famille :** Apprentissage par renforcement, contrôle hors politique (off-policy). **Objectif :** Apprendre la valeur d'une action dans un état afin de maximiser la récompense future.

**Intuition et fonctionnement.** L'agent observe un état, choisit une action, reçoit récompense et nouvel état, puis met à jour Q vers la meilleure valeur future possible.

**Formule ou critère essentiel :**  
`Q(s,a)←Q(s,a)+α[r+γ max_{a'}Q(s',a')−Q(s,a)]`

**Définition des symboles :** s : état ; a : action ; r : récompense ; s' : état suivant ; α : taux d'apprentissage ; γ : facteur d'actualisation.  
**Hyperparamètres et paramètres appris :** alpha ; gamma ; epsilon ; nombre d'épisodes.

**Deux avantages :**
- Apprend une politique sans modèle de transition préalable.
- Converge sous certaines conditions de couverture et de taux d'apprentissage.

**Deux limites :**
- Table Q devient volumineuse avec de grands espaces d'états.
- Apprentissage sensible à l'exploration et à la stochasticité.

### Schéma explicatif

Le schéma original ci-dessous représente le cheminement conceptuel de l'algorithme : données/état en entrée, opération d'apprentissage, puis sortie. Les étiquettes décrivent le mécanisme spécifique de cette méthode.

### Contextes d'utilisation

**Cas d'usage 1 :** Optimiser une séquence d'actions de recouvrement dans un simulateur validé (finance).  
**Cas d'usage 2 :** Optimiser les déplacements d'un robot en entrepôt.

**Données et prérequis :** Apprentissage par renforcement, contrôle hors politique (off-policy). Il faut disposer de variables cohérentes avec la tâche, traiter les valeurs manquantes, choisir un encodage adapté et, lorsque la méthode dépend des distances ou des marges, mettre les variables à l'échelle. Les hyperparamètres doivent être réglés sur validation, jamais sur le test final.

**Cas peu approprié :** Un tableau statique sans états, actions, transitions et récompenses observées.  
**Métriques adaptées :** Récompense cumulée, taux de réussite, convergence sur plusieurs graines.

### Analyse de l'expérience Python

Le code entraîne Q-learning sur FrozenLake-v1 non glissant pendant 4 000 épisodes ; il affiche la politique greedy extraite de Q. Les transitions viennent de l'environnement Gymnasium, pas du fichier Excel ; la politique dépend des paramètres et doit être évaluée sur plusieurs épisodes/graines.



```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(10, 2.4))
ax.set_xlim(0, 3); ax.set_ylim(0, 1); ax.axis("off")
etapes = ['État s', 'Action, récompense, mise à jour Q', 'Politique optimale']
for j, texte in enumerate(etapes):
    ax.text(j+0.5, 0.52, texte, ha="center", va="center", wrap=True,
            bbox=dict(boxstyle="round,pad=0.6", facecolor="#e8eef7", edgecolor="#24476b"))
    if j < 2:
        ax.annotate("", xy=(j+0.93,0.52), xytext=(j+0.67,0.52),
                    arrowprops=dict(arrowstyle="->", lw=1.8))
ax.set_title("27. Q-learning — schéma de principe", fontsize=12)
plt.show()
```

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** SARSA (State–Action–Reward–State–Action). **Famille :** Apprentissage par renforcement, contrôle on-policy. **Objectif :** Apprendre une valeur d'action selon la politique effectivement suivie, y compris son exploration.

**Intuition et fonctionnement.** L'agent choisit a, observe r et s', choisit ensuite a' selon sa politique, puis met à jour Q(s,a) à partir de Q(s',a').

**Formule ou critère essentiel :**  
`Q(s,a)←Q(s,a)+α[r+γQ(s',a')−Q(s,a)]`

**Définition des symboles :** s,a,r,s',a' : transition ; α : taux d'apprentissage ; γ : facteur d'actualisation.  
**Hyperparamètres et paramètres appris :** alpha ; gamma ; epsilon ; nombre d'épisodes.

**Deux avantages :**
- Prend en compte les actions réellement suivies.
- Peut être plus prudent sous exploration que Q-learning.

**Deux limites :**
- Dépend fortement de la politique d'exploration.
- Peut apprendre plus lentement.

### Schéma explicatif

Le schéma original ci-dessous représente le cheminement conceptuel de l'algorithme : données/état en entrée, opération d'apprentissage, puis sortie. Les étiquettes décrivent le mécanisme spécifique de cette méthode.

### Contextes d'utilisation

**Cas d'usage 1 :** Apprendre une politique de relance sous contraintes de risque (simulateur financier).  
**Cas d'usage 2 :** Planifier un robot en entrepôt avec pénalité de collision.

**Données et prérequis :** Apprentissage par renforcement, contrôle on-policy. Il faut disposer de variables cohérentes avec la tâche, traiter les valeurs manquantes, choisir un encodage adapté et, lorsque la méthode dépend des distances ou des marges, mettre les variables à l'échelle. Les hyperparamètres doivent être réglés sur validation, jamais sur le test final.

**Cas peu approprié :** Des données tabulaires fixes sans boucle d'interaction ni récompenses.  
**Métriques adaptées :** Récompense cumulée, taux de réussite, stabilité entre graines.

### Analyse de l'expérience Python

Le code applique SARSA à FrozenLake-v1 et met à jour Q avec l'action suivante réellement choisie. Il affiche une politique ; ce résultat relève d'une simulation Gymnasium et ne doit pas être décrit comme une expérience sur les clients de la base.



```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(10, 2.4))
ax.set_xlim(0, 3); ax.set_ylim(0, 1); ax.axis("off")
etapes = ['État s et action a', "Transition et Q(s',a')", 'Politique suivie']
for j, texte in enumerate(etapes):
    ax.text(j+0.5, 0.52, texte, ha="center", va="center", wrap=True,
            bbox=dict(boxstyle="round,pad=0.6", facecolor="#e8eef7", edgecolor="#24476b"))
    if j < 2:
        ax.annotate("", xy=(j+0.93,0.52), xytext=(j+0.67,0.52),
                    arrowprops=dict(arrowstyle="->", lw=1.8))
ax.set_title("28. SARSA (State–Action–Reward–State–Action) — schéma de principe", fontsize=12)
plt.show()
```

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

## 29. DQN — cible conceptuelle

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Deep Q-Network (DQN). **Famille :** Apprentissage par renforcement profond. **Objectif :** Approcher la fonction Q par un réseau neuronal pour traiter des espaces d'états plus grands.

**Intuition et fonctionnement.** L'agent sélectionne des actions par epsilon-greedy, stocke les transitions dans une mémoire de rejeu, entraîne un réseau sur des mini-lots et utilise un réseau cible pour stabiliser la cible de Bellman.

**Formule ou critère essentiel :**  
`y=r+γ(1−terminé) max_{a'}Q_cible(s',a')`

**Définition des symboles :** r : récompense ; γ : actualisation ; Q_cible : réseau cible ; s' : état suivant ; action maximisante.  
**Hyperparamètres et paramètres appris :** Architecture du réseau ; learning rate ; gamma ; epsilon ; replay buffer ; batch size ; fréquence de mise à jour cible.

**Deux avantages :**
- Traite des espaces d'états complexes.
- Combine apprentissage profond et Q-learning.

**Deux limites :**
- Coûteux en données et calcul.
- Instabilité et sensibilité aux hyperparamètres.

### Schéma explicatif

**Cas d'usage 2 :** Apprendre le contrôle d'un robot mobile en entrepôt.

**Données et prérequis :** Apprentissage par renforcement profond. Les hyperparamètres doivent être réglés sur validation, jamais sur le test final.

**Cas peu approprié :** Un simple fichier Excel sans environnement dynamique ni récompenses séquentielles.  
**Métriques adaptées :** Récompense cumulée, taux de réussite, courbes d'apprentissage et variabilité entre graines.

### Analyse de l'expérience Python
