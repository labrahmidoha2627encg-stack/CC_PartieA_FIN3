## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Régression linéaire (Linear Regression). **Famille :** Régression supervisée. **Objectif :** Prédire une valeur quantitative par une combinaison linéaire des variables.

**Intuition et fonctionnement.** Le modèle reçoit X et y, estime les coefficients par moindres carrés, puis calcule une prédiction pour chaque nouvelle observation.

**Formule ou critère essentiel :**  
`ŷᵢ = β₀ + Σⱼ βⱼxᵢⱼ ; J(β)=1/(2n)Σᵢ(yᵢ−ŷᵢ)²`

**Définition des symboles :** ŷᵢ : prédiction ; β₀ : constante ; βⱼ : coefficients appris ; xᵢⱼ : valeur de la variable j ; yᵢ : valeur réelle ; n : nombre d'observations.  
**Hyperparamètres et paramètres appris :** fit_intercept ; paramètres appris : β₀ et βⱼ.

**Deux avantages :**
- Simple et rapide.
- Coefficients relativement interprétables.

**Deux limites :**
- Ne capture pas spontanément les relations non linéaires.
- Sensible aux valeurs extrêmes et à la colinéarité.

### Schéma explicatif

Le schéma original ci-dessous représente le cheminement conceptuel de l'algorithme : données/état en entrée, opération d'apprentissage, puis sortie. Les étiquettes décrivent le mécanisme spécifique de cette méthode.

### Contextes d'utilisation

**Cas d'usage 1 :** Prévoir le chiffre d'affaires selon les dépenses commerciales (gestion).  
**Cas d'usage 2 :** Estimer les coûts logistiques selon distance, volume et carburant.

**Données et prérequis :** Régression supervisée. Il faut disposer de variables cohérentes avec la tâche, traiter les valeurs manquantes, choisir un encodage adapté et, lorsque la méthode dépend des distances ou des marges, mettre les variables à l'échelle. Les hyperparamètres doivent être réglés sur validation, jamais sur le test final.

**Cas peu approprié :** Classer un client en défaut/non-défaut : la cible est une catégorie, pas une quantité continue.  
**Métriques adaptées :** MAE, RMSE et R².

### Analyse de l'expérience Python

Le code utilise Xr_ent/Xr_test et une cible numérique synthétique ; LinearRegression() apprend une droite/hyperplan et affiche la MAE. Cela illustre l'ajustement et l'erreur absolue, mais ne fournit aucun résultat valable sur le dataset de crédit et ne permet pas d'évaluer une classification.



```python
# Schéma original : ajustement d'une droite aux observations
rng = np.random.default_rng(42)
x = np.linspace(0, 10, 35); y = 2.2*x + 4 + rng.normal(0, 3, len(x))
coef = np.polyfit(x, y, 1); xx = np.linspace(0, 10, 100)
fig, ax = plt.subplots(figsize=(8, 4.2))
ax.scatter(x, y, s=35, alpha=.75, label="Observations (X, y)")
ax.plot(xx, np.polyval(coef, xx), lw=2.8, label="Droite ajustée : ŷ = β₀ + β₁X")
for xi, yi in zip(x[::5], y[::5]): ax.vlines(xi, yi, np.polyval(coef, xi), color="gray", alpha=.5, ls="--")
ax.set(title="Régression linéaire : minimiser les résidus", xlabel="Variable explicative X", ylabel="Cible y")
ax.legend(); ax.grid(alpha=.2); plt.show()
```

**Comment lire le schéma :** cette représentation illustre le mécanisme central de l'algorithme 1. Les valeurs et points dessinés sont pédagogiques : ils servent à expliquer le principe et ne constituent pas les résultats de l'expérience Python qui suit.

```python
from sklearn.linear_model import LinearRegression
modele=LinearRegression().fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Ridge (Ridge Regression). **Famille :** Régression supervisée pénalisée L2. **Objectif :** Réduire l'amplitude des coefficients pour stabiliser une régression.

**Intuition et fonctionnement.** Le modèle calcule une prédiction linéaire et minimise l'erreur quadratique augmentée d'une pénalité sur les carrés des coefficients.

**Formule ou critère essentiel :**  
`J(β)=||y−Xβ||²₂/n + αΣⱼβⱼ²`

**Définition des symboles :** X : matrice des variables ; y : cible quantitative ; βⱼ : coefficients appris ; α : intensité de régularisation.  
**Hyperparamètres et paramètres appris :** alpha ; fit_intercept ; paramètres appris : coefficients β.

**Deux avantages :**
- Stabilise les coefficients en présence de variables corrélées.
- Réduit le surapprentissage par régularisation.

**Deux limites :**
- Ne supprime généralement pas les variables.
- Le choix d'alpha influence fortement le compromis biais-variance.

### Schéma explicatif

Le schéma original ci-dessous représente le cheminement conceptuel de l'algorithme : données/état en entrée, opération d'apprentissage, puis sortie. Les étiquettes décrivent le mécanisme spécifique de cette méthode.

### Contextes d'utilisation

**Cas d'usage 1 :** Prévoir des revenus avec indicateurs macroéconomiques corrélés (finance).  
**Cas d'usage 2 :** Prévoir les dépenses d'approvisionnement avec plusieurs facteurs liés.

**Données et prérequis :** Régression supervisée pénalisée L2. Il faut disposer de variables cohérentes avec la tâche, traiter les valeurs manquantes, choisir un encodage adapté et, lorsque la méthode dépend des distances ou des marges, mettre les variables à l'échelle. Les hyperparamètres doivent être réglés sur validation, jamais sur le test final.

**Cas peu approprié :** Une cible catégorielle de défaut, car Ridge est une régression et non un classifieur.  
**Métriques adaptées :** MAE, RMSE, R².

### Analyse de l'expérience Python

Le code standardise les entrées, ajuste Ridge(alpha=1.0) et affiche MAE sur un exemple de régression synthétique. Il montre l'effet de la pénalisation L2, mais aucune valeur imprimée ne doit être attribuée à la base de crédit.



```python
# Schéma original : Ridge contracte les coefficients sans les annuler en général
noms = ["Prix", "Revenu", "Âge", "Ancienneté", "Taux"]
beta = np.array([2.8, -2.1, 1.6, .9, -.5]); ridge = beta * np.array([.66,.70,.62,.73,.58])
pos=np.arange(len(noms)); fig, ax=plt.subplots(figsize=(8,4.2))
ax.barh(pos+.18, beta, height=.32, label="Régression sans pénalité", alpha=.8)
ax.barh(pos-.18, ridge, height=.32, label="Ridge (L2)", alpha=.9)
ax.axvline(0,color="black",lw=.8); ax.set_yticks(pos,noms); ax.invert_yaxis()
ax.set(title="Ridge : réduire l'amplitude des coefficients", xlabel="Valeur du coefficient")
ax.legend(); ax.grid(axis="x",alpha=.2); plt.show()
```

**Comment lire le schéma :** cette représentation illustre le mécanisme central de l'algorithme 2. Les valeurs et points dessinés sont pédagogiques : ils servent à expliquer le principe et ne constituent pas les résultats de l'expérience Python qui suit.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Lasso (Lasso Regression). **Famille :** Régression supervisée pénalisée L1. **Objectif :** Effectuer une régression tout en pouvant annuler certains coefficients.

**Intuition et fonctionnement.** Après mise à l'échelle, le modèle minimise l'erreur quadratique et la somme des valeurs absolues des coefficients ; certains deviennent exactement nuls.

**Formule ou critère essentiel :**  
`J(β)=||y−Xβ||²₂/(2n)+αΣⱼ|βⱼ|`

**Définition des symboles :** βⱼ : coefficient appris ; α : intensité de la pénalité L1 ; n : nombre d'observations.  
**Hyperparamètres et paramètres appris :** alpha ; max_iter ; paramètres appris : βⱼ.

**Deux avantages :**
- Effectue une sélection intégrée de variables.
- Produit un modèle parfois plus parcimonieux.

**Deux limites :**
- Peut sélectionner arbitrairement une variable parmi des variables corrélées.
- Résultats sensibles à alpha et à l'échelle.

### Schéma explicatif

Le schéma original ci-dessous représente le cheminement conceptuel de l'algorithme : données/état en entrée, opération d'apprentissage, puis sortie. Les étiquettes décrivent le mécanisme spécifique de cette méthode.

### Contextes d'utilisation

**Cas d'usage 1 :** Sélectionner des indicateurs financiers pour prévoir une marge.  
**Cas d'usage 2 :** Identifier les facteurs associés aux coûts de transport.

**Données et prérequis :** Régression supervisée pénalisée L1. Il faut disposer de variables cohérentes avec la tâche, traiter les valeurs manquantes, choisir un encodage adapté et, lorsque la méthode dépend des distances ou des marges, mettre les variables à l'échelle. Les hyperparamètres doivent être réglés sur validation, jamais sur le test final.

**Cas peu approprié :** Prédire directement une classe binaire avec LassoRegressor plutôt qu'un classifieur adapté.  
**Métriques adaptées :** MAE, RMSE, R² ; nombre de coefficients non nuls.

### Analyse de l'expérience Python

Le code standardise puis ajuste Lasso(alpha=0.1, max_iter=10000) et affiche R² sur la cible numérique synthétique. Il illustre la pénalisation L1 et la parcimonie ; il ne mesure pas la performance de défaut de crédit.



```python
# Schéma original : Lasso peut rendre certains coefficients exactement nuls
noms=["Prix", "Revenu", "Âge", "Publicité", "Saison", "Distance"]
coef=np.array([1.8,0,-.9,0,1.2,0]); fig,ax=plt.subplots(figsize=(8,4.2))
ax.bar(noms,coef); ax.axhline(0,color="black",lw=.8)
for i,v in enumerate(coef): ax.text(i, v + (.08 if v>=0 else -.16), "0 : écartée" if v==0 else f"{v:.1f}",ha="center",fontsize=9)
ax.set(title="Lasso : sélection automatique de variables (L1)", ylabel="Coefficient estimé")
ax.grid(axis="y",alpha=.2); plt.show()
```

**Comment lire le schéma :** cette représentation illustre le mécanisme central de l'algorithme 3. Les valeurs et points dessinés sont pédagogiques : ils servent à expliquer le principe et ne constituent pas les résultats de l'expérience Python qui suit.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Régression polynomiale (Polynomial Regression). **Famille :** Régression supervisée. **Objectif :** Modéliser une relation courbe en ajoutant des puissances et interactions des variables.

**Intuition et fonctionnement.** Les variables sont transformées en termes polynomiaux ; un modèle linéaire est ensuite ajusté sur ces nouvelles colonnes.

**Formule ou critère essentiel :**  
`ŷ=β₀+β₁x+β₂x²+…+β_dxᵈ`

**Définition des symboles :** x : variable d'entrée ; d : degré choisi ; β : coefficients appris ; ŷ : prédiction.  
**Hyperparamètres et paramètres appris :** degree ; include_bias ; alpha du Ridge éventuellement utilisé.

**Deux avantages :**
- Représente des courbes simples.
- S'appuie sur des modèles linéaires connus.

**Deux limites :**
- Le nombre de termes augmente vite avec les variables et le degré.
- Un degré élevé peut surapprendre et extrapoler mal.

### Schéma explicatif

Le schéma original ci-dessous représente le cheminement conceptuel de l'algorithme : données/état en entrée, opération d'apprentissage, puis sortie. Les étiquettes décrivent le mécanisme spécifique de cette méthode.

### Contextes d'utilisation

**Cas d'usage 1 :** Modéliser une relation non linéaire entre taux et demande de crédit (finance).  
**Cas d'usage 2 :** Prévoir une demande logistique évoluant de façon courbe selon le prix.

**Données et prérequis :** Régression supervisée. Il faut disposer de variables cohérentes avec la tâche, traiter les valeurs manquantes, choisir un encodage adapté et, lorsque la méthode dépend des distances ou des marges, mettre les variables à l'échelle. Les hyperparamètres doivent être réglés sur validation, jamais sur le test final.

**Cas peu approprié :** Un problème de classification à nombreuses variables et interactions complexes sans contrôle de complexité.  
**Métriques adaptées :** MAE, RMSE, R².

### Analyse de l'expérience Python

Le code crée des caractéristiques de degré 2, standardise puis ajuste Ridge(alpha=10) ; il affiche MAE sur une cible numérique synthétique. La sortie illustre une courbe approximée, sans conclusion sur les défauts binaires.



```python
# Schéma original : les puissances de X permettent d'ajuster une courbe
rng=np.random.default_rng(4); x=np.linspace(-3,3,45); y=.65*x**2-.7*x+1+rng.normal(0,1.1,len(x)); xx=np.linspace(-3,3,200)
fig,ax=plt.subplots(figsize=(8,4.2)); ax.scatter(x,y,s=28,alpha=.65,label="Observations")
for d,style in [(1,"--"),(2,"-"),(5,":")]:
    c=np.polyfit(x,y,d); ax.plot(xx,np.polyval(c,xx),style,lw=2.4,label=f"Degré {d}")
ax.set(title="Régression polynomiale : de la droite à la courbe",xlabel="X",ylabel="y"); ax.legend(); ax.grid(alpha=.2); plt.show()
```

**Comment lire le schéma :** cette représentation illustre le mécanisme central de l'algorithme 4. Les valeurs et points dessinés sont pédagogiques : ils servent à expliquer le principe et ne constituent pas les résultats de l'expérience Python qui suit.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Arbre de régression (Decision Tree Regressor). **Famille :** Régression supervisée. **Objectif :** Prédire une valeur numérique en partitionnant l'espace des variables en régions.

**Intuition et fonctionnement.** À chaque nœud, l'arbre choisit une variable et un seuil qui réduisent l'erreur quadratique ; les feuilles prédisent une moyenne locale.

### Présentation

**Hyperparamètres et paramètres appris :** epsilon ; règle de décroissance ; horizon ; méthode de mise à jour.

**Deux avantages :**
- Très simple à implémenter.
- Garantit une exploration minimale si ε>0.

- Exploration non dirigée et parfois inefficace.
- Choix sensible au réglage de ε.

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Q-learning. **Famille :** Apprentissage par renforcement, contrôle hors politique (off-policy). **Objectif :** Apprendre la valeur d'une action dans un état afin de maximiser la récompense future.

**Intuition et fonctionnement.** L'agent observe un état, choisit une action, reçoit récompense et nouvel état, puis met à jour Q vers la meilleure valeur future possible.

**Formule ou critère essentiel :**  
`Q(s,a)←Q(s,a)+α[r+γ max_{a'}Q(s',a')−Q(s,a)]`

**Définition des symboles :** s : état ; a : action ; r : récompense ; s' : état suivant ; α : taux d'apprentissage ; γ : facteur d'actualisation.  
**Hyperparamètres et paramètres appris :** alpha ; gamma ; epsilon ; nombre d'épisodes.

**Deux avantages :**
- Apprend une politique sans modèle de transition préalable.
- Converge sous certaines conditions de couverture et de taux d'apprentissage.

**Deux limites :**
- Table Q devient volumineuse avec de grands espaces d'états.
- Apprentissage sensible à l'exploration et à la stochasticité.

### Schéma explicatif

Le schéma original ci-dessous représente le cheminement conceptuel de l'algorithme : données/état en entrée, opération d'apprentissage, puis sortie. Les étiquettes décrivent le mécanisme spécifique de cette méthode.

### Contextes d'utilisation

**Cas d'usage 1 :** Optimiser une séquence d'actions de recouvrement dans un simulateur validé (finance).  
**Cas d'usage 2 :** Optimiser les déplacements d'un robot en entrepôt.

**Données et prérequis :** Apprentissage par renforcement, contrôle hors politique (off-policy). Il faut disposer de variables cohérentes avec la tâche, traiter les valeurs manquantes, choisir un encodage adapté et, lorsque la méthode dépend des distances ou des marges, mettre les variables à l'échelle. Les hyperparamètres doivent être réglés sur validation, jamais sur le test final.

**Cas peu approprié :** Un tableau statique sans états, actions, transitions et récompenses observées.  
**Métriques adaptées :** Récompense cumulée, taux de réussite, convergence sur plusieurs graines.

### Analyse de l'expérience Python

Le code entraîne Q-learning sur FrozenLake-v1 non glissant pendant 4 000 épisodes ; il affiche la politique greedy extraite de Q. Les transitions viennent de l'environnement Gymnasium, pas du fichier Excel ; la politique dépend des paramètres et doit être évaluée sur plusieurs épisodes/graines.



```python
# Schéma original : Q-learning met à jour la meilleure valeur future possible
fig,ax=plt.subplots(figsize=(8,4.5));ax.axis("off");ax.set_xlim(0,10);ax.set_ylim(0,6)
ax.text(1.4,3.2,"État s",ha="center",va="center",bbox=dict(boxstyle="round",fc="#e8f1fb"))
ax.text(3.8,3.2,"Action a\n+ récompense r",ha="center",va="center",bbox=dict(boxstyle="round",fc="#fff2cc"))
ax.text(6.4,3.2,"Nouvel état s′",ha="center",va="center",bbox=dict(boxstyle="round",fc="#e8f1fb"))
ax.text(8.8,3.2,"Mise à jour\nQ(s,a)",ha="center",va="center",bbox=dict(boxstyle="round",fc="#d9ead3"))
for x1,x2 in [(2,3),(4.9,5.5),(7.3,8)]:ax.annotate("",(x2,3.2),(x1,3.2),arrowprops=dict(arrowstyle="->",lw=1.8))
ax.text(5,.9,"Cible : r + γ maxₐ′ Q(s′, a′)   (hors politique)",ha="center",fontsize=12)
ax.set_title("Q-learning : apprendre à partir du meilleur futur estimé");plt.show()
```

**Comment lire le schéma :** cette représentation illustre le mécanisme central de l'algorithme 27. Les valeurs et points dessinés sont pédagogiques : ils servent à expliquer le principe et ne constituent pas les résultats de l'expérience Python qui suit.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

## 28. SARSA

### Présentation

**Nom :** SARSA (State–Action–Reward–State–Action). **Famille :** Apprentissage par renforcement, contrôle on-policy. **Objectif :** Apprendre une valeur d'action selon la politique effectivement suivie, y compris son exploration.

**Formule ou critère essentiel :**  
`Q(s,a)←Q(s,a)+α[r+γQ(s',a')−Q(s,a)]`

**Définition des symboles :** s,a,r,s',a' : transition ; α : taux d'apprentissage ; γ : facteur d'actualisation.  
**Hyperparamètres et paramètres appris :** alpha ; gamma ; epsilon ; nombre d'épisodes.

**Deux avantages :**
- Prend en compte les actions réellement suivies.
- Peut être plus prudent sous exploration que Q-learning.

**Deux limites :**
- Dépend fortement de la politique d'exploration.
- Peut apprendre plus lentement.

### Schéma explicatif

### Contextes d'utilisation

**Données et prérequis :** Apprentissage par renforcement, contrôle on-policy. Il faut disposer de variables cohérentes avec la tâche, traiter les valeurs manquantes, choisir un encodage adapté et, lorsque la méthode dépend des distances ou des marges, mettre les variables à l'échelle. Les hyperparamètres doivent être réglés sur validation, jamais sur le test final.

**Cas peu approprié :** Des données tabulaires fixes sans boucle d'interaction ni récompenses.  
**Métriques adaptées :** Récompense cumulée, taux de réussite, stabilité entre graines.

### Analyse de l'expérience Python

Le code applique SARSA à FrozenLake-v1 et met à jour Q avec l'action suivante réellement choisie. Il affiche une politique ; ce résultat relève d'une simulation Gymnasium et ne doit pas être décrit comme une expérience sur les clients de la base.



```python
# Schéma original : SARSA utilise l'action réellement choisie à l'étape suivante
fig,ax=plt.subplots(figsize=(8.5,4.5));ax.axis("off");ax.set_xlim(0,10);ax.set_ylim(0,6)
labels=[(1,"Sₜ"),(3,"Aₜ"),(5,"Rₜ₊₁"),(7,"Sₜ₊₁"),(9,"Aₜ₊₁")]
for x,t in labels:ax.text(x,3.2,t,ha="center",va="center",fontsize=14,bbox=dict(boxstyle="circle,pad=.55",fc="#e8f1fb" if x in [1,7] else "#fff2cc"))
for (x,_),(xx,__) in zip(labels[:-1],labels[1:]):ax.annotate("",(xx-.45,3.2),(x+.45,3.2),arrowprops=dict(arrowstyle="->",lw=1.7))
ax.text(5,1.2,"Cible : Rₜ₊₁ + γ Q(Sₜ₊₁, Aₜ₊₁)",ha="center",fontsize=12,bbox=dict(boxstyle="round",fc="#d9ead3"))
ax.set_title("SARSA : la mise à jour suit la trajectoire réellement exécutée");plt.show()
```

**Comment lire le schéma :** cette représentation illustre le mécanisme central de l'algorithme 28. Les valeurs et points dessinés sont pédagogiques : ils servent à expliquer le principe et ne constituent pas les résultats de l'expérience Python qui suit.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

**Intuition et fonctionnement.** L'agent sélectionne des actions par epsilon-greedy, stocke les transitions dans une mémoire de rejeu, entraîne un réseau sur des mini-lots et utilise un réseau cible pour stabiliser la cible de Bellman.

**Formule ou critère essentiel :**  
`y=r+γ(1−terminé) max_{a'}Q_cible(s',a')`

**Définition des symboles :** r : récompense ; γ : actualisation ; Q_cible : réseau cible ; s' : état suivant ; action maximisante.  
**Hyperparamètres et paramètres appris :** Architecture du réseau ; learning rate ; gamma ; epsilon ; replay buffer ; batch size ; fréquence de mise à jour cible.

**Deux avantages :**
- Traite des espaces d'états complexes.
- Combine apprentissage profond et Q-learning.

**Deux limites :**
- Coûteux en données et calcul.
- Instabilité et sensibilité aux hyperparamètres.

### Schéma explicatif

Le schéma original ci-dessous représente le cheminement conceptuel de l'algorithme : données/état en entrée, opération d'apprentissage, puis sortie. Les étiquettes décrivent le mécanisme spécifique de cette méthode.

### Contextes d'utilisation

**Cas d'usage 1 :** Optimiser une politique séquentielle dans un simulateur de gestion de risque (finance).  
**Cas d'usage 2 :** Apprendre le contrôle d'un robot mobile en entrepôt.

**Données et prérequis :** Apprentissage par renforcement profond. Il faut disposer de variables cohérentes avec la tâche, traiter les valeurs manquantes, choisir un encodage adapté et, lorsque la méthode dépend des distances ou des marges, mettre les variables à l'échelle. Les hyperparamètres doivent être réglés sur validation, jamais sur le test final.

## Q6 — Exploration et préparation des données

La préparation suit les principes suivants :

1. Vérifier les dimensions, les types, les valeurs manquantes et les doublons.
2. Vérifier l'identifiant et l'équilibre de la variable cible.
3. Exclure `ID`, qui identifie les lignes mais n'apporte pas d'information généralisable sur le risque.
4. Conserver les modalités inhabituelles des variables codées, en les signalant plutôt que de leur attribuer une signification non documentée.
5. Séparer les données en ensembles entraînement (60 %), validation (20 %) et test final (20 %) avec stratification.
6. Apprendre les imputations, l'encodage et la standardisation uniquement sur l'entraînement, au moyen de pipelines `scikit-learn`, afin d'éviter la fuite d'information.

Les valeurs extrêmes des montants de crédit, factures et paiements ne sont pas supprimées automatiquement : elles peuvent représenter de vrais profils financiers. Elles sont repérées par une règle IQR à titre descriptif seulement.

```python
# Diagnostic de qualité
print("Valeurs manquantes :", int(credit.isna().sum().sum()))
print("Lignes dupliquées :", int(credit.duplicated().sum()))
print("Identifiants dupliqués :", int(credit['ID'].duplicated().sum()))
print("\nTypes :")
display(credit.dtypes.value_counts().rename_axis("Type").to_frame("Nombre de colonnes"))
print("\nRépartition de la cible :")
display(credit[TARGET].value_counts().rename(index={0:"Pas de défaut",1:"Défaut"}).to_frame("Effectif"))
display((credit[TARGET].value_counts(normalize=True)*100).round(2).rename(index={0:"Pas de défaut",1:"Défaut"}).to_frame("Pourcentage"))

# Diagnostic des valeurs atypiques selon l'IQR (sans suppression automatique)
numeric_cols = credit.select_dtypes(include=np.number).columns.drop(['ID', TARGET], errors='ignore')
outlier_rows = []
for col in numeric_cols:
    q1, q3 = credit[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    low, high = q1 - 1.5*iqr, q3 + 1.5*iqr
    outlier_rows.append((col, int(((credit[col] < low) | (credit[col] > high)).sum()), float(low), float(high)))
outlier_summary = pd.DataFrame(outlier_rows, columns=["Variable","Nombre hors IQR","Borne basse","Borne haute"])
display(outlier_summary.sort_values("Nombre hors IQR", ascending=False).head(12))

credit[TARGET].value_counts(normalize=True).sort_index().plot(kind="bar")
plt.title("Répartition de la variable cible")
plt.xlabel("Défaut le mois suivant (0 = non, 1 = oui)")
plt.ylabel("Proportion")
plt.tight_layout()
plt.show()
```

```python
# Préparation des variables et découpage sans fuite d'information
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

X = credit.drop(columns=[TARGET, 'ID']).copy()
y = credit[TARGET].astype(int).copy()

# Variables catégorielles codées ; les statuts PAY_* sont conservés comme variables numériques ordinales.
categorical_features = ['SEX', 'EDUCATION', 'MARRIAGE']
numeric_features = [c for c in X.columns if c not in categorical_features]

X_trainval, X_test, y_trainval, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)
X_train, X_val, y_train, y_val = train_test_split(
    X_trainval, y_trainval, test_size=0.25, random_state=RANDOM_STATE, stratify=y_trainval
)

numeric_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
categorical_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])
preprocessor = ColumnTransformer([
    ('num', numeric_pipe, numeric_features),
    ('cat', categorical_pipe, categorical_features)
])

print("Entraînement :", X_train.shape, "| Validation :", X_val.shape, "| Test :", X_test.shape)
print("Taux de défaut — train :", round(y_train.mean(), 4),
      "| validation :", round(y_val.mean(), 4), "| test :", round(y_test.mean(), 4))
```

## Q7 — Expérimentations supervisées : classification du défaut

La classification est l'utilisation principale et la plus directement justifiée par la fiche UCI. Le jeu est déséquilibré (la classe défaut est minoritaire). L'accuracy seule serait donc insuffisante. Les indicateurs retenus sont la précision, le rappel, le F1, ROC-AUC et PR-AUC. En gestion du risque, le rappel de la classe défaut est important, mais il doit être mis en regard de la précision et du coût des faux positifs.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier, HistGradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, average_precision_score, confusion_matrix,
                             ConfusionMatrixDisplay, classification_report)

# Référence simple : prédire la classe majoritaire.
majority_pred = np.zeros(len(y_test), dtype=int)
baseline = {
    "Modèle": "Référence majoritaire",
    "Accuracy": accuracy_score(y_test, majority_pred),
    "Précision": precision_score(y_test, majority_pred, zero_division=0),
    "Rappel": recall_score(y_test, majority_pred, zero_division=0),
    "F1": f1_score(y_test, majority_pred, zero_division=0),
    "ROC-AUC": np.nan, "PR-AUC": np.nan
}

classifiers = {
    "Régression logistique": LogisticRegression(max_iter=1200, class_weight='balanced', random_state=RANDOM_STATE),
    "KNN": KNeighborsClassifier(n_neighbors=15, weights='distance'),
    "Naive Bayes": GaussianNB(),
    "Arbre de décision": DecisionTreeClassifier(max_depth=8, min_samples_leaf=30, class_weight='balanced', random_state=RANDOM_STATE),
    "Forêt aléatoire": RandomForestClassifier(n_estimators=150, max_depth=12, min_samples_leaf=10,
                                               class_weight='balanced_subsample', n_jobs=-1, random_state=RANDOM_STATE),
    "SVM RBF": SVC(C=1.0, kernel='rbf', probability=True, class_weight='balanced', random_state=RANDOM_STATE),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=100, learning_rate=0.05, max_depth=2, random_state=RANDOM_STATE),
    "AdaBoost": AdaBoostClassifier(n_estimators=100, learning_rate=0.5, random_state=RANDOM_STATE)
}

# XGBoost est ajouté si la bibliothèque est disponible.
try:
    from xgboost import XGBClassifier
    classifiers["XGBoost"] = XGBClassifier(
        n_estimators=150, max_depth=3, learning_rate=0.05,
        subsample=0.85, colsample_bytree=0.85, eval_metric='logloss',
        random_state=RANDOM_STATE, n_jobs=-1
    )
except ImportError:
    print("XGBoost non installé : modèle ignoré. Pour l'ajouter : !pip install xgboost")

fitted_classifiers = {}
validation_rows = []
for name, estimator in classifiers.items():
    # KNN est coûteux sur 18 000 lignes : apprentissage sur un sous-échantillon stratifié.
    if name == "KNN":
        X_fit, _, y_fit, _ = train_test_split(
            X_train, y_train, train_size=min(8000, len(X_train)),
            random_state=RANDOM_STATE, stratify=y_train
        )
    else:
        X_fit, y_fit = X_train, y_train

    pipe = Pipeline([('prep', preprocessor), ('model', estimator)])
    pipe.fit(X_fit, y_fit)
    val_prob = pipe.predict_proba(X_val)[:, 1]
    val_pred = (val_prob >= 0.50).astype(int)
    validation_rows.append({
        "Modèle": name,
        "Précision": precision_score(y_val, val_pred, zero_division=0),
        "Rappel": recall_score(y_val, val_pred, zero_division=0),
        "F1": f1_score(y_val, val_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_val, val_prob),
        "PR-AUC": average_precision_score(y_val, val_prob)
    })
    fitted_classifiers[name] = pipe

validation_results = pd.DataFrame(validation_rows).sort_values("PR-AUC", ascending=False)
display(validation_results.round(3))
```

```python
# Évaluation finale : le meilleur modèle est choisi sur validation (PR-AUC),
# puis évalué une seule fois sur le jeu de test.
best_name = validation_results.iloc[0]["Modèle"]
best_model = fitted_classifiers[best_name]
test_prob = best_model.predict_proba(X_test)[:, 1]
test_pred = (test_prob >= 0.50).astype(int)

test_metrics = pd.DataFrame([{
    "Modèle retenu sur validation": best_name,
    "Accuracy": accuracy_score(y_test, test_pred),
    "Précision": precision_score(y_test, test_pred, zero_division=0),
    "Rappel": recall_score(y_test, test_pred, zero_division=0),
    "F1": f1_score(y_test, test_pred, zero_division=0),
    "ROC-AUC": roc_auc_score(y_test, test_prob),
    "PR-AUC": average_precision_score(y_test, test_prob)
}])
display(test_metrics.round(3))
print(classification_report(y_test, test_pred, target_names=["Pas de défaut", "Défaut"], zero_division=0))
ConfusionMatrixDisplay.from_predictions(y_test, test_pred, display_labels=["Pas de défaut","Défaut"])
plt.title(f"Matrice de confusion — {best_name}")
plt.tight_layout()
plt.show()
```

## Q7 — Régression : estimation numérique de la cible binaire (usage secondaire)

Les algorithmes de régression linéaire, Ridge, Lasso, polynomiale, arbre, forêt aléatoire et Gradient Boosting sont conçus pour une cible continue. Le dataset UCI ne propose pas de cible continue de risque observé : sa cible est binaire (0/1). Pour pouvoir illustrer ces méthodes avec le fichier fourni, on leur fait estimer une valeur numérique de la cible binaire, puis on interprète cette valeur comme un score brut.

**Limite essentielle :** ce protocole n'est pas équivalent à une régression sur une véritable variable financière continue, et les sorties des régressions ne sont pas nécessairement des probabilités calibrées ou bornées entre 0 et 1. Les métriques MAE, RMSE et R² sont calculées sur le test, mais elles ne remplacent pas l'évaluation probabiliste d'un modèle de classification.

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import PolynomialFeatures
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

regressors = {
    "Régression linéaire": LinearRegression(),
    "Ridge": Ridge(alpha=10.0),
    "Lasso": Lasso(alpha=0.01, max_iter=3000, random_state=RANDOM_STATE),
    "Arbre de régression": DecisionTreeRegressor(max_depth=8, min_samples_leaf=30, random_state=RANDOM_STATE),
    "Forêt aléatoire": RandomForestRegressor(n_estimators=120, max_depth=12, min_samples_leaf=10,
                                              n_jobs=-1, random_state=RANDOM_STATE),
    "Gradient Boosting": GradientBoostingRegressor(n_estimators=100, learning_rate=0.05,
                                                    max_depth=2, random_state=RANDOM_STATE)
}
# Régression polynomiale de degré 2 sur un sous-ensemble de variables pour limiter l'explosion dimensionnelle.
poly_features = ['LIMIT_BAL', 'AGE', 'PAY_0', 'PAY_2', 'PAY_AMT1', 'BILL_AMT1']
poly_pre = ColumnTransformer([
    ('poly_num', Pipeline([('imp', SimpleImputer(strategy='median')),
                           ('poly', PolynomialFeatures(degree=2, include_bias=False)),
                           ('scale', StandardScaler())]), poly_features)
], remainder='drop')
regression_rows = []

for name, estimator in regressors.items():
    pipe = Pipeline([('prep', preprocessor), ('model', estimator)])
    pipe.fit(X_train, y_train)
    pred = pipe.predict(X_test)
    regression_rows.append({
        "Modèle": name,
        "MAE": mean_absolute_error(y_test, pred),
        "RMSE": np.sqrt(mean_squared_error(y_test, pred)),
        "R²": r2_score(y_test, pred)
    })

poly_model = Pipeline([('prep', poly_pre), ('model', LinearRegression())])
poly_model.fit(X_train[poly_features], y_train)
poly_pred = poly_model.predict(X_test[poly_features])
regression_rows.append({
    "Modèle": "Régression polynomiale degré 2",
    "MAE": mean_absolute_error(y_test, poly_pred),
    "RMSE": np.sqrt(mean_squared_error(y_test, poly_pred)),
    "R²": r2_score(y_test, poly_pred)
})

regression_results = pd.DataFrame(regression_rows).sort_values("RMSE")
display(regression_results.round(3))
```

## Q7 — Apprentissage non supervisé : profils clients et anomalies

Pour le clustering, la cible de défaut et l'identifiant sont exclus de l'apprentissage : les algorithmes doivent regrouper les clients sans connaître leur étiquette. Un échantillon est utilisé pour limiter le coût de calcul. La silhouette mesure la cohésion/séparation géométrique ; elle ne prouve pas qu'un segment est utile commercialement. Les étiquettes de défaut ne sont consultées qu'après le clustering, à titre d'évaluation externe descriptive.

Les variables de montants sont standardisées. Les variables `SEX`, `EDUCATION` et `MARRIAGE` sont encodées en one-hot. Les statuts de remboursement restent numériques ordinaux.

```python
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score, adjusted_rand_score

# Transformation apprise sur l'entraînement uniquement ; l'échantillon provient du jeu train.
X_train_transformed = preprocessor.fit_transform(X_train)
X_cluster = X_train_transformed[:5000]
y_cluster = y_train.iloc[:5000].to_numpy()

cluster_models = {
    "K-means (k=2)": KMeans(n_clusters=2, n_init=10, random_state=RANDOM_STATE),
    "Classification hiérarchique (k=2)": AgglomerativeClustering(n_clusters=2),
    "Mélanges gaussiens (k=2)": GaussianMixture(n_components=2, covariance_type='diag', random_state=RANDOM_STATE)
}
cluster_rows = []
cluster_labels = {}
for name, model in cluster_models.items():
    labels = model.fit_predict(X_cluster)
    cluster_labels[name] = labels
    sil = silhouette_score(X_cluster, labels) if len(np.unique(labels)) > 1 else np.nan
    ari = adjusted_rand_score(y_cluster, labels)
    cluster_rows.append({"Méthode": name, "Silhouette": sil, "ARI (externe)": ari,
                         "Nombre de groupes": len(set(labels))})

# DBSCAN : paramètres à ajuster selon l'échelle et la densité ; les points -1 sont des bruits.
dbscan = DBSCAN(eps=2.5, min_samples=15, n_jobs=-1)
db_labels = dbscan.fit_predict(X_cluster)
cluster_labels["DBSCAN"] = db_labels
valid = db_labels != -1
db_sil = silhouette_score(X_cluster[valid], db_labels[valid]) if valid.sum() > 1 and len(set(db_labels[valid])) > 1 else np.nan
cluster_rows.append({"Méthode":"DBSCAN", "Silhouette (hors bruit)":db_sil,
                     "ARI (externe)":adjusted_rand_score(y_cluster, db_labels),
                     "Nombre de groupes":len(set(db_labels) - {-1})})

cluster_results = pd.DataFrame(cluster_rows)
display(cluster_results.round(3))
```

```python
# Profil des segments : après l'apprentissage non supervisé seulement,
# on examine la proportion de défaut dans chaque groupe.
for name, labels in cluster_labels.items():
    profile = pd.DataFrame({"cluster": labels, "default": y_cluster})
    print("\n", name)
    display(profile.groupby("cluster").agg(
        effectif=("default","size"),
        taux_defaut=("default","mean")
    ).round(3))
```

## Q7 — Réduction de dimension : ACP, t-SNE et UMAP

La réduction de dimension est réalisée sur les variables explicatives préparées. Pour les visualisations t-SNE et UMAP, un échantillon est utilisé car ces méthodes peuvent être coûteuses. La couleur des points représente la cible uniquement pour faciliter la lecture après projection ; elle n'est pas utilisée pour calculer la projection.

- **ACP** : la variance expliquée est une mesure intrinsèque utile.
- **t-SNE** : la projection sert à visualiser des voisinages locaux ; les distances globales entre amas ne doivent pas être interprétées comme des distances exactes.
- **UMAP** : la conservation des voisinages dépend des paramètres et de la graine ; l'installation est facultative.

```python
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

# ACP sur les données d'entraînement transformées
pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca = pca.fit_transform(X_train_transformed)
print("Variance expliquée par les deux premiers axes :", np.round(pca.explained_variance_ratio_, 4))
print("Variance cumulée :", round(pca.explained_variance_ratio_.sum(), 4))

# Échantillon reproductible pour les cartes t-SNE / UMAP
rng = np.random.RandomState(RANDOM_STATE)
idx = rng.choice(len(X_train_transformed), size=min(1500, len(X_train_transformed)), replace=False)
X_vis = X_train_transformed[idx]
y_vis = y_train.iloc[idx].to_numpy()

tsne = TSNE(n_components=2, perplexity=30, init='pca', learning_rate='auto',
            random_state=RANDOM_STATE)
X_tsne = tsne.fit_transform(X_vis)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].scatter(X_pca[:,0], X_pca[:,1], c=y_train, s=8, alpha=.5)
axes[0].set_title("ACP — train")
axes[1].scatter(X_tsne[:,0], X_tsne[:,1], c=y_vis, s=8, alpha=.5)
axes[1].set_title("t-SNE — échantillon train")
for ax in axes:
    ax.set_xlabel("Dimension 1")
    ax.set_ylabel("Dimension 2")
plt.tight_layout()
plt.show()

# UMAP est facultatif ; le notebook reste exécutable si le paquet n'est pas installé.
try:
    import umap
    reducer = umap.UMAP(n_components=2, random_state=RANDOM_STATE)
    X_umap = reducer.fit_transform(X_vis)
    plt.figure(figsize=(7,5))
    plt.scatter(X_umap[:,0], X_umap[:,1], c=y_vis, s=8, alpha=.5)
    plt.title("UMAP — échantillon train")
    plt.xlabel("Dimension 1"); plt.ylabel("Dimension 2")
    plt.tight_layout(); plt.show()
except ImportError:
    print("UMAP non installé. Pour l'ajouter dans Colab : !pip install umap-learn")
```

## Q7 — Détection d'anomalies : Isolation Forest

Isolation Forest est ajusté uniquement sur les variables explicatives de l'entraînement, sans utiliser la cible. Les observations signalées comme anomalies sont ensuite comparées aux étiquettes de défaut pour une évaluation externe indicative. Un défaut de paiement n'est pas nécessairement une anomalie statistique et une anomalie statistique n'est pas nécessairement un défaut.

Le paramètre `contamination` représente ici une proportion supposée d'observations atypiques (valeur illustrative de 5 %), et non le taux de défaut observé.

```python
from sklearn.ensemble import IsolationForest
from sklearn.metrics import precision_score, recall_score, f1_score

iso = IsolationForest(n_estimators=200, contamination=0.05, random_state=RANDOM_STATE, n_jobs=-1)
iso.fit(X_train_transformed)

X_test_transformed = preprocessor.transform(X_test)
# IsolationForest : -1 = anomalie, 1 = normal
anomaly_pred = (iso.predict(X_test_transformed) == -1).astype(int)

anomaly_summary = pd.DataFrame([{
    "Proportion signalée comme anomalie": anomaly_pred.mean(),
    "Précision vs défaut": precision_score(y_test, anomaly_pred, zero_division=0),
    "Rappel vs défaut": recall_score(y_test, anomaly_pred, zero_division=0),
    "F1 vs défaut": f1_score(y_test, anomaly_pred, zero_division=0)
}])
display(anomaly_summary.round(3))
print("Attention : ces métriques comparent une anomalie non supervisée à la cible défaut ; elles ne valident pas une détection de fraude.")
```

## Q7 — Apprentissage par renforcement : environnements Gymnasium

Les algorithmes de bandits, Q-learning, SARSA et DQN **ne doivent pas être entraînés sur le fichier CSV/XLS des clients**. Ce fichier est un tableau statique, alors que l'apprentissage par renforcement nécessite une interaction agent–environnement et des trajectoires de récompenses.

Pour Q-learning et SARSA, l'exemple ci-dessous utilise `FrozenLake-v1`, un environnement public de Gymnasium. Le DQN du notebook initial reste conceptuel : une implémentation complète n'est pas obligatoire selon les consignes, mais cette limite doit être indiquée. Gymnasium ne fournit pas un benchmark classique de bandit multi-bras dans ses environnements de base ; le code epsilon-greedy de la partie A doit donc être présenté comme une démonstration distincte, et non comme une expérimentation sur les données bancaires.

```python
# Environnement public Gymnasium pour Q-learning et SARSA
# Dépendance : !pip install gymnasium
try:
    import gymnasium as gym

    def train_tabular(env_name="FrozenLake-v1", algo="q_learning",
                      episodes=5000, alpha=0.1, gamma=0.99, epsilon=0.1, seed=42):
        env = gym.make(env_name, is_slippery=False)
        n_states = env.observation_space.n
        n_actions = env.action_space.n
        Q = np.zeros((n_states, n_actions), dtype=float)
        returns = []

        for ep in range(episodes):
            state, _ = env.reset(seed=seed + ep)
            done = False
            total_reward = 0
            while not done:
                if np.random.random() < epsilon:
                    action = env.action_space.sample()
                else:
                    action = int(np.argmax(Q[state]))

                next_state, reward, terminated, truncated, _ = env.step(action)
                done = terminated or truncated

                if algo == "q_learning":
                    target = reward + (0 if done else gamma * np.max(Q[next_state]))
                elif algo == "sarsa":
                    if np.random.random() < epsilon:
                        next_action = env.action_space.sample()
                    else:
                        next_action = int(np.argmax(Q[next_state]))
                    target = reward + (0 if done else gamma * Q[next_state, next_action])
                else:
                    raise ValueError("algo doit être 'q_learning' ou 'sarsa'")

                Q[state, action] += alpha * (target - Q[state, action])
                state = next_state
                total_reward += reward
            returns.append(total_reward)

        env.close()
        return Q, np.array(returns)

    q_table, q_returns = train_tabular(algo="q_learning")
    sarsa_table, sarsa_returns = train_tabular(algo="sarsa")
    print("Récompense moyenne Q-learning (100 derniers épisodes) :", q_returns[-100:].mean())
    print("Récompense moyenne SARSA (100 derniers épisodes) :", sarsa_returns[-100:].mean())

    plt.figure(figsize=(8,4))
    plt.plot(pd.Series(q_returns).rolling(100).mean(), label="Q-learning")
    plt.plot(pd.Series(sarsa_returns).rolling(100).mean(), label="SARSA")
    plt.xlabel("Épisode"); plt.ylabel("Récompense moyenne glissante (100 épisodes)")
    plt.title("Apprentissage sur FrozenLake-v1")
    plt.legend(); plt.tight_layout(); plt.show()
except ImportError:
    print("Gymnasium non installé. Dans Colab, exécuter : !pip install gymnasium")
```
